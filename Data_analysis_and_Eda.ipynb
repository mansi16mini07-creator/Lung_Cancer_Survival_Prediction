#### UNIVARIATE ANALYSIS

```python
# Numerical feature distributions.

# Numerical feature distributions
df.select_dtypes(include=np.number).hist(figsize=(14, 10),bins=20)
plt.tight_layout()
plt.show()
```

```python
# categorical feature 
df.select_dtypes(include='str').nunique()
```

```python
categorical_cols = data.select_dtypes(
    include=["str", "category", "bool"]
).columns

for col in categorical_cols:
    print(f"\n{col}")
    print(data[col].value_counts(dropna=False).head(15))
```

```python
# Bivariate analysis of features against survived.

if "gender" in df.columns:
    sns.countplot(data=df, x="gender", hue="survived")
    plt.title("Survival Outcome by Gender")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Correlation analysis.

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# Outlier detection.
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=data[col])
    plt.title(f"Outlier Check: {col}")
    plt.show()
```

# Key observations and conclusions.

1. Dataset structure: The dataset contains patient records and multiple features that can be explored to understand lung cancer survival outcomes.
2. Data quality: Missing values and duplicate records were examined to determine whether data cleaning was required.
3. Survival distribution: The distribution of the target variable, survived, was analyzed to identify whether the dataset has class imbalance.
4. Feature distribution: Numerical and categorical features were examined using descriptive statistics, histograms, and frequency counts to understand their distributions.
5. Feature-target relationships: Individual features were compared with survived to identify variables that may be associated with survival outcomes.
6. Outlier analysis: Box plots were used to identify unusual numerical values that may require further investigation.
7. Correlation analysis: A correlation matrix was used to examine relationships among numerical variables and identify potentially redundant features.

### Data preprocessing decisions for machine learning.

1. Keep survived separate from input features and verify its class meanings.
2. Convert columns to appropriate numeric, categorical, or date types.
3. Remove identifiers and features that leak the outcome or are unavailable at prediction time.
4. Split the data before fitting preprocessing steps; use stratification for classification when appropriate.
5. Apply StandardScaler when appropriate, especially for Logistic Regression.
6. Use One-Hot Encoding for nominal categories and suitable ordinal encoding for genuinely ordered categories.
